```python
import numpy as np
import pandas as pd

# ==========================================
# 第四步：数据清洗与转换
# ==========================================

# 1. 查缺失值（看看哪列有空缺）
print("=== 缺失值检查 ===")
print(geo_df.isnull().sum())

# 2. 处理缺失值（我们用 fillna(0) 填空缺练习一下）
# 注意：这里为了不破坏原来的 geo_df，我们赋值给一个新的变量 geo_df_filled
geo_df_filled = geo_df.fillna(0)

# 3. 数学转换 (Log2 转换)
# 必须先剔除第一列的文字（ID_REF），只留数字，然后再加1求对数
data_only = geo_df_filled.iloc[:, 1:] 
log2_data = np.log2(data_only + 1)

# 看一眼转换后的结果（只显示前3行）
print("\n=== Log2 转换后的数据 ===")
print(log2_data.head(3))

# 4. 数据排序 (按 GSM15965 这一列的表达量从高到低排)
# 因为 log2_data 已经没有 ID_REF 了，我们直接按 GSM15965 排
log2_sorted = log2_data.sort_values(by='GSM15965', ascending=False)
print("\n=== 排序后的前5个高表达基因 ===")
print(log2_sorted.head())

# ==========================================
# 第五步：数据保存（打完收工）
# ==========================================

# 1. 存为 CSV（极度重要：index=False 不要把前面的行号存进去）
log2_sorted.to_csv('GSE1009_log2_sorted.csv', index=False)
print("\n=== 数据已成功保存为 CSV 文件！ ===")

# 2. 存为 Excel（可选，如果你想试试也行，这里我帮你注释掉，节省时间）
# log2_sorted.to_excel('GSE1009_log2_sorted.xlsx', index=False)
```
